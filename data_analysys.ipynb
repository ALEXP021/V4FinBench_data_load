```python
# python 3.12.7
# -*- coding: utf-8 -*-
"""
Notebook do przeglądania przygotowanych danych
Analiza plików Parquet po przetwarzaniu wstępnym
"""

import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

print("="*70)
print("📊 ANALIZA PRZYGOTOWANYCH DANYCH")
print("="*70)
```

```python
# --- 1. Ścieżka do danych ---
DATA_DIR = r"C:\Users\afina\Documents\Inforshare\Final project\data"  # Podaj ścieżkę do folderu danych
PROCESSED_DIR = os.path.join(DATA_DIR, "processed")
SELECTED_HORIZON = "h1"

print(f"\n📁 Folder z przygotowanymi danymi: {PROCESSED_DIR}")
print(f"📊 Horyzont prognozowania: {SELECTED_HORIZON}")

# Sprawdzamy, czy pliki istnieją
files_to_check = [
    f'X_train_{SELECTED_HORIZON}.parquet',
    f'y_train_{SELECTED_HORIZON}.parquet',
    f'X_test_{SELECTED_HORIZON}.parquet',
    f'y_test_{SELECTED_HORIZON}.parquet',
    f'preprocessor_{SELECTED_HORIZON}.pkl'
]

print("\n🔍 Sprawdzanie dostępności plików:")
for file in files_to_check:
    file_path = os.path.join(PROCESSED_DIR, file)
    exists = os.path.exists(file_path)
    size = os.path.getsize(file_path) / (1024**2) if exists else 0
    status = "✅" if exists else "❌"
    if exists:
        print(f"  {status} {file}: {size:.2f} MB")
    else:
        print(f"  {status} {file}: NIE ZNALEZIONO")
```

```python
# --- 2. Wczytywanie danych ---
print("\n" + "="*70)
print("📥 WCZYTYWANIE DANYCH")
print("="*70)

print("\n🔄 Wczytywanie zbioru treningowego (X_train)...")
X_train = pd.read_parquet(os.path.join(PROCESSED_DIR, f'X_train_{SELECTED_HORIZON}.parquet'))
print(f"✅ X_train wczytany: {X_train.shape}")

print("\n🔄 Wczytywanie zmiennej docelowej (y_train)...")
y_train = pd.read_parquet(os.path.join(PROCESSED_DIR, f'y_train_{SELECTED_HORIZON}.parquet'))
print(f"✅ y_train wczytany: {y_train.shape}")

print("\n🔄 Wczytywanie zbioru testowego (X_test)...")
X_test = pd.read_parquet(os.path.join(PROCESSED_DIR, f'X_test_{SELECTED_HORIZON}.parquet'))
print(f"✅ X_test wczytany: {X_test.shape}")

print("\n🔄 Wczytywanie zmiennej docelowej (y_test)...")
y_test = pd.read_parquet(os.path.join(PROCESSED_DIR, f'y_test_{SELECTED_HORIZON}.parquet'))
print(f"✅ y_test wczytany: {y_test.shape}")
```

```python
# --- 3. Ogólne informacje o danych ---
print("\n" + "="*70)
print("📊 OGÓLNE INFORMACJE")
print("="*70)

print(f"\n📌 Rozmiary zbiorów:")
print(f"  X_train: {X_train.shape[0]:,} wierszy, {X_train.shape[1]} cech")
print(f"  y_train: {y_train.shape[0]:,} wierszy")
print(f"  X_test:  {X_test.shape[0]:,} wierszy, {X_test.shape[1]} cech")
print(f"  y_test:  {y_test.shape[0]:,} wierszy")

print(f"\n📌 Łączna liczba:")
print(f"  Zbiór treningowy: {X_train.shape[0] + y_train.shape[0]:,} przykładów")
print(f"  Zbiór testowy: {X_test.shape[0] + y_test.shape[0]:,} przykładów")
print(f"  Razem: {X_train.shape[0] + y_train.shape[0] + X_test.shape[0] + y_test.shape[0]:,} przykładów")
```

```python
# --- 4. Анализ целевой переменной ---
print("\n" + "="*70)
print("🎯 АНАЛИЗ ЦЕЛЕВОЙ ПЕРЕМЕННОЙ (main_label)")
print("="*70)

# Для обучающей выборки
train_counts = y_train['main_label'].value_counts()
train_pct = y_train['main_label'].value_counts(normalize=True) * 100

print(f"\n📊 Обучающая выборка:")
print(f"  Класс 0 (здоровые): {train_counts[0]:,} ({train_pct[0]:.2f}%)")
print(f"  Класс 1 (банкротство): {train_counts[1]:,} ({train_pct[1]:.2f}%)")
print(f"  Дисбаланс: {train_pct[1]:.2f}% банкротств")

# Для тестовой выборки
test_counts = y_test['main_label'].value_counts()
test_pct = y_test['main_label'].value_counts(normalize=True) * 100

print(f"\n📊 Тестовая выборка:")
print(f"  Класс 0 (здоровые): {test_counts[0]:,} ({test_pct[0]:.2f}%)")
print(f"  Класс 1 (банкротство): {test_counts[1]:,} ({test_pct[1]:.2f}%)")
print(f"  Дисбаланс: {test_pct[1]:.2f}% банкротств")
```

```python
# --- 4. Analiza zmiennej docelowej ---
print("\n" + "="*70)
print("🎯 ANALIZA ZMIENNEJ DOCELOWEJ (main_label)")
print("="*70)

# Dla zbioru treningowego
train_counts = y_train['main_label'].value_counts()
train_pct = y_train['main_label'].value_counts(normalize=True) * 100

print(f"\n📊 Zbiór treningowy:")
print(f"  Klasa 0 (zdrowe): {train_counts[0]:,} ({train_pct[0]:.2f}%)")
print(f"  Klasa 1 (bankructwo): {train_counts[1]:,} ({train_pct[1]:.2f}%)")
print(f"  Dysbalans: {train_pct[1]:.2f}% bankructw")

# Dla zbioru testowego
test_counts = y_test['main_label'].value_counts()
test_pct = y_test['main_label'].value_counts(normalize=True) * 100

print(f"\n📊 Zbiór testowy:")
print(f"  Klasa 0 (zdrowe): {test_counts[0]:,} ({test_pct[0]:.2f}%)")
print(f"  Klasa 1 (bankructwo): {test_counts[1]:,} ({test_pct[1]:.2f}%)")
print(f"  Dysbalans: {test_pct[1]:.2f}% bankructw")
```

```python
# --- 5. Analiza braków danych ---
print("\n" + "="*70)
print("🔍 ANALIZA BRAKÓW DANYCH")
print("="*70)

# Sprawdzamy brakujące wartości w X_train
null_counts = X_train.isnull().sum()
null_cols = null_counts[null_counts > 0].index.tolist()

print(f"\n📊 Brakujące wartości w X_train:")
if len(null_cols) > 0:
    print(f"  Znaleziono {len(null_cols)} cech z brakującymi wartościami:")
    for col in null_cols[:10]:  # Pokazujemy pierwsze 10
        print(f"    {col}: {null_counts[col]:,} ({null_counts[col]/len(X_train)*100:.2f}%)")
    if len(null_cols) > 10:
        print(f"    ... i {len(null_cols)-10} kolejnych")
else:
    print("  ✅ Brak brakujących wartości")

print(f"\n📊 Brakujące wartości w X_test:")
null_counts_test = X_test.isnull().sum()
null_cols_test = null_counts_test[null_counts_test > 0].index.tolist()
if len(null_cols_test) > 0:
    print(f"  Znaleziono {len(null_cols_test)} cech z brakującymi wartościami")
else:
    print("  ✅ Brak brakujących wartości")

# --- 6. Podgląd pierwszych wierszy ---
print("\n" + "="*70)
print("👀 PODGLĄD DANYCH")
print("="*70)

print(f"\n📊 Pierwsze 5 wierszy X_train:")
print(X_train.head())

print(f"\n📊 Pierwsze 10 wartości y_train:")
print(y_train.head(10))

print(f"\n📊 Pierwsze 5 wierszy X_test:")
print(X_test.head())

print(f"\n📊 Pierwsze 10 wartości y_test:")
print(y_test.head(10))

# --- 7. Analiza korelacji (wybrane cechy) ---
print("\n" + "="*70)
print("🔍 ANALIZA KORELACJI")
print("="*70)

# Bierzemy pierwsze 20 cech do analizy
sample_cols = X_train.columns[:20]
correlation_matrix = X_train[sample_cols].corr()

print(f"\n📊 Macierz korelacji dla pierwszych 20 cech:")
print("  Rozmiar macierzy:", correlation_matrix.shape)
print("\n  Pierwsze 5x5 korelacji:")
print(correlation_matrix.iloc[:5, :5])

# Silne korelacje
print("\n📊 Silne korelacje (|r| > 0.8):")
strong_corr = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_val = correlation_matrix.iloc[i, j]
        if abs(corr_val) > 0.8:
            strong_corr.append((correlation_matrix.columns[i], correlation_matrix.columns[j], corr_val))

if strong_corr:
    print(f"  Znaleziono {len(strong_corr)} silnych korelacji:")
    for col1, col2, corr in strong_corr[:10]:  # Pokazujemy tylko pierwsze 10
        print(f"    {col1} ↔ {col2}: {corr:.3f}")
else:
    print("  ✅ Brak silnych korelacji (>0.8)")

# --- 8. Zapis statystyk do pliku ---
print("\n" + "="*70)
print("💾 ZAPIS STATYSTYK")
print("="*70)

# Zapisujemy podstawowe statystyki
stats = {
    'X_train_shape': X_train.shape,
    'X_test_shape': X_test.shape,
    'y_train_counts': train_counts.to_dict(),
    'y_test_counts': test_counts.to_dict(),
    'feature_count': X_train.shape[1],
    'train_bankruptcy_rate': train_pct[1],
    'test_bankruptcy_rate': test_pct[1],
    'null_count': len(null_cols)
}

stats_df = pd.DataFrame([stats])
stats_path = os.path.join(PROCESSED_DIR, f'data_statistics_{SELECTED_HORIZON}.csv')
stats_df.to_csv(stats_path, index=False)
print(f"✅ Statystyki zapisane: {stats_path}")
```
